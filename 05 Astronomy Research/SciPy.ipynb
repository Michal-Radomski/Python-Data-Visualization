```python
import numpy as np
from scipy import stats, optimize, integrate
import matplotlib.pyplot as plt
```

```python
data1 = np.random.normal(0, 1, 100)
data2 = np.random.normal(0.5, 1, 100)
t_stat, p_value = stats.ttest_ind(data1, data2)
print(f"T-stat: {t_stat:.2f}, P-value: {p_value:.4f}")
```

```python
def f(x): return np.sin(x) * np.exp(-x)
result, error = integrate.quad(f, 0, np.pi)
print(f"Integral: {result:.4f}, Error: {error:.2e}")
```

```python
x = np.linspace(0, 10, 50)
y = 2.5 * np.sin(x) + np.random.normal(0, 0.3, 50)
popt, _ = optimize.curve_fit(lambda x, a, b: a * np.sin(x) + b, x, y)
plt.plot(x, y, 'o', label='Data')
plt.plot(x, popt[0] * np.sin(x) + popt[1], '-', label='Fit')
plt.legend()
plt.show()
```
